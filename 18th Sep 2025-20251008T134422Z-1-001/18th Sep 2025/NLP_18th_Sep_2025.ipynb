# Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
text = ["I like food food food food", "I do not like food",'NLP is good for text data']
```

```python
vectorizer = CountVectorizer(binary=True)
```

```python
nlp = vectorizer.fit_transform(text)
nlp
```

```python
nlp.shape
```

```python
import pandas as pd
pd.DataFrame(nlp.toarray(),columns=vectorizer.get_feature_names_out())
```

```python
vectorizer.vocabulary_
```

# n-grams

```python
text = ["I like food", "I do not like food",'NLP is good for text data']
```

```python
tf_idf = TfidfVectorizer(binary=True, ngram_range=(1, 4))
vec_rep = tf_idf.fit_transform(text)
pd.DataFrame(vec_rep.toarray(),columns=tf_idf.get_feature_names_out())
```

# Stemming and Lemmatization

```python
from nltk.stem import PorterStemmer # stemming
from nltk.stem import WordNetLemmatizer # lemmatization
```

```python
text = "Natural Language Processing is a field of artificial intelligence that focuses on the interaction between computers and human language."
text
```

```python
list_of_words = text.split(" ")
list_of_words
```

```python
tokens = word_tokenize(text)
tokens
```

```python
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in tokens]
print(tokens)
print("******************")
print(stemmed_words)
```

# Lemmatization

```python
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
lemmatizer_words = [lemmatizer.lemmatize(word) for word in tokens]
print(tokens)
print("******************")
print(lemmatizer_words)
```
