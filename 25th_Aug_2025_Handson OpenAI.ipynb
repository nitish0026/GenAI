```python
prompt = "Give me a numbered list of all the modules one should study in Data Science? use apprx 1000 words."
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role":"user",
         "content":prompt}       

    ]
)

print(response.choices[0].message.content)
```

```python
prompt = "Give me a numbered list of all the modules one should study in Data Science?"
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role":"user",
         "content":prompt}       

    ],
    max_tokens=100
)

print(response.choices[0].message.content)
```

```python
prompt = "Give me a numbered list of all the modules one should study in Data Science? use apprx 1000 words."
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role":"user",
         "content":prompt}       

    ],
    stop="10."
)

print(response.choices[0].message.content)
```

```python
# Completions API (LEGACY) - used for
### sentiment analysis, Summarization, information extraction, NER etc...

prompt = "There was a king, who had "
response = client.completions.create(
    model="gpt-4o-mini",
    prompt=prompt, max_tokens=1000)
print(response.choices[0].text)
```

# creating Embeddings

# Image - Dall-e 2 and 3

```python
response = client.images.generate(
    model = "dall-e-2",
    prompt = "A digital image captured of an astronaut who is driving a school bus in a garden.",
    size = "1024x1024",
    
    n=1)
response
```

```python
response = client.images.generate(
    model = "dall-e-3",
    prompt = "A digital image captured of an astronaut who is driving a school bus in a garden.",
    size = "1024x1024",
    quality="hd",
    n=1)
image_url = response.data[0].url
print(image_url)
```
