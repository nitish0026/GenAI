# Loop Function

#### break - continue - and pass

```python
# break statement

for i in range(1,20):
    if i==5:
        break
    print(i)
```

```python
# continue statement
for i in range(1,20):
    if i == 10:
        continue
    print(i)
    print("Hello Python")
```

```python
# pass statement

for i in range(1,20):
    if i == 10:
        pass
    print(i)
```

```python
for i in range(20,0,-5):
    print(i)
```

```python
for i in "BANGALORE":
    print(i)
```
