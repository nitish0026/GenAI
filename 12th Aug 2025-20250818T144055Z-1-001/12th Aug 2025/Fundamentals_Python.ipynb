```python
email = input("enter your email id here :")
password = input("enter your password here :")

if email =="kumar.s@learnbay.co" and password =='welcome123':
  print("Welcome to LearnBay")
elif email =="kumar.s@learnbay.co" and password !='welcome123':
  print("Incorrect Password")
  password = input("enter your password again :")
  if password =='welcome123':
    print("Finally correct")
  else:
    print("Still incorrect")
else:
  print("Incorrect credentials")
```

# Loops in Python

```python
# while
number = int(input("Enter the number here : "))

i = 1

while i <20:
  print(number, "*", i, "=", number * i)
  i = i+1
```

```python
x = 1
while x < 5:
  print(x)
  x +=1

else:
  print("limit crossed")
```

```python
# For loop function

for i in {1,2,3,4,5}:
  print(i)
```

```python
def is_even_number(num):
  if type(num) == int:
    if num % 2 == 0:
      return "Number is even"
    else:
      return "Number is odd"
  else:
    return "Please enter integer value"
```

```python
is_even_number(2307/3456)
```

```python
is_even_number(0)
```
