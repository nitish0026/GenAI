# Setup API Key

```python
from langchain_openai import OpenAI
```

```python
f = open("keys/.openai_api_key.txt")
OPENAI_API_KEY = f.read()
```

# Step 1 - Logic 1 (ChatPromptTemplate)

```python
from langchain_core.prompts import ChatPromptTemplate
prompt_template = ChatPromptTemplate.from_messages([
    ("system","You are a helpful AI Tutor with expertise in Data Science and Artificial Intelligence."),
    ("human", "what is {topic}?")])
```

```python
prompt_template.input_variables
```

# Step 2 - Model

```python
from langchain_openai import ChatOpenAI
chat_model = ChatOpenAI(openai_api_key = OPENAI_API_KEY)
prompt = "What is Feature Engineering?"
print(chat_model.invoke(prompt))
```

# Step 3 - logic 2 - Output Parser

```python
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
```

# Step 4 - Chain

```python
chain = prompt_template | chat_model | output_parser

# Response
input = {"topic": "Generative AI"}
chain.invoke(input)
```

```python
from IPython.display import display, Markdown
display(Markdown(chain.invoke(input)))
```
