```python
# First agent - gather the information basis the problem statement
strategy_creator = autogen.AssistantAgent(name="strategy_creator", 
                                          llm_config=strategy_config_list,
                                          system_message=""" You are a logical marketing manager tasked with crafting a
                                          marketing strategy. Gether details on the product, target audience, and budget.
                                          Then identify the best marketing channels which will get the product maximum
                                          visibility.""",
                                         )
```

```python
content_config_list = {"config_list":[{"model":"gpt-4o-mini"}]}
```

```python
# 2nd Agent - create content basis the given input by first agent

content_creator = autogen.AssistantAgent(name="content_creator", 
                                          llm_config=content_config_list,
                                          system_message=""" You are a Content Executor, a creative AI Agent.
                                          Your task is to craft engaging, channel-specific content that alings
                                           with the marketing strategy, ensuring it captures attention and 
                                           resonates with the target audience.""",
                                         )
```

# Proxy agent

```python
user_proxy = autogen.UserProxyAgent(
    name="User",
    human_input_mode="ALWAYS",
     is_termination_msg=lambda x:x.get("content", "") and x.get("content","").rstrip().endswith("TERMINATE"),
    code_execution_config=False,)
```

```python
product_details = [
"""Water bottle for school childern under the age of 16 years. The marketing budget is 2000 USD""",
]
```

```python
writing_task = ["""Write an engaging content for the mentioned product and channel"""]
```

# Initiate the Sequential Conversation among them

```python
chat_results = autogen.initiate_chats(
    [
        {"sender": user_proxy,
         "recipient":strategy_creator,
         "message": product_details[0],
         "clear_history":True,
         "silent":False,
         "summary_method":"last_msg",},
        {"sender": user_proxy,
         "recipient":content_creator,
         "message": writing_task[0],
         "summary_method":"last_msg",},]
)
```
